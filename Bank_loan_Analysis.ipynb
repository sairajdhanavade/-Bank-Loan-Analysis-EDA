### Total loan Application


```python
tla = df["id"].count()
print("Total Loan Application:",tla)
```

### Total MTD Application

```python
latest_date= df.issue_date.max()
latest_date_year = latest_date.year
latest_date_month = latest_date.month
mtd_data = df[(df["issue_date"].dt.year == latest_date_year) & (df["issue_date"].dt.month == latest_date_month)]
tmtda = mtd_data["id"].count()
print("Total MTD Applications:",tmtda)
```

### Total Funded Amount

```python
tfa = df["loan_amount"].sum()
print("Total Funded Amount:",tfa)
```

### Total MTD Funded Amount

```python
tmtdfa = mtd_data["loan_amount"].sum()
print("Total MTD Funded Amount::",tmtdfa)
```

### Total Amount Received

```python
tfr = df["total_payment"].sum()
print("Total Amount received:",tfa)
```

### Total MTD Amount Rececived

```python
tmtdfr = mtd_data["total_payment"].sum()
print("Total MTD Funded Amount::",tmtdfr)
```

### Average Interest Rate

```python
air = df["int_rate"].mean()
print("Average Interest Rate:",round(air,4))
```

### Average Debt-to-Income Ratio (DTI)

```python
adti = df["dti"].mean()
print("Average Debt-to-Income Ratio (DTI):",round(adti,4))
```

### Good Loan Metrics

```python
Good_Loan = df[df["loan_status"].isin(["Current","Fully Paid"])]
Good_Loan
```

```python
glac = Good_Loan["id"].count()
print("Total Good Loan Applications:",glac)
glfa = Good_Loan["loan_amount"].sum()
print("Total Good Loan Funded Amount:",glfa)
glra = Good_Loan["total_payment"].sum()
print("Total Good Loan Rececived Amount:",glra)
glp=(glac/tla)*100
print("Percentage of good loan:",round(glp,2))
```

### Bad Loan Matrics

```python
Bad_Loan = df[df["loan_status"]=="Charged Off"]
Bad_Loan
```

```python
blac = Bad_Loan["id"].count()
print("Total Bad Loan Applications:",blac)
blfa = Bad_Loan["loan_amount"].sum()
print("Total Bad Loan Funded Amount:",blfa)
blra = Bad_Loan["total_payment"].sum()
print("Total Bad Loan Rececived Amount:",blra)
blp=(blac/tla)*100
print("Percentage of Bad Loan:",round(blp,2))
```

### Monthly Trends Total Funded Amount

```python
df1 = df.sort_values("issue_date")
plt.figure(figsize=(15,5))
sns.lineplot(data=df1,x=df1.issue_date.dt.month_name(),y="loan_amount",ci=False,color = "green")
plt.title("Total Funded Amount by Month",fontsize=15)
plt.xlabel("Month")
plt.ylabel("Funded Amount")
plt.tight_layout()
plt.show()
```

### Monthly Trends  for Total Amount Received

```python
plt.figure(figsize=(15,5))
sns.lineplot(data=df1,x=df1.issue_date.dt.month_name(),y="total_payment",ci=False)
plt.title("Total Recevied Amount by Month",fontsize=15)
plt.xlabel("Month")
plt.ylabel("Received Amount")
plt.tight_layout()
plt.show()
```

### Count Total Loan Application per Months

```python
plt.figure(figsize=(15,5))
ax=sns.countplot(data=df1,x=df1.issue_date.dt.month_name())
plt.bar_label(ax.containers[0])
plt.show()
```

### Loan Term Analysis by Total Funded Amount


```python
plt.figure(figsize=(10,5))
plt.pie(df.groupby("term")["loan_amount"].sum(),labels=(df.groupby("term")["loan_amount"].sum()).index,
        autopct='%1.1f%%',startangle=90)
plt.title("Total Funded Amount by Term")
plt.tight_layout()
plt.show()
```
